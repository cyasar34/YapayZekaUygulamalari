## 1. NumPy — Sayılarla Hızlı Çalışmak

**NumPy**, Python'da sayı dizileriyle hızlı ve toplu işlem yapmayı sağlayan kütüphanedir.
Görüntüler, ses ve model çıktıları bilgisayarda **sayı dizileri** olarak durur; YZ kütüphanelerinin hepsi (pandas, scikit-learn, TensorFlow) NumPy üzerine kuruludur.

Evrensel kısaltması `np`'dir.

````python
import numpy as np

liste = [10, 20, 30]
dizi = np.array(liste)    # listeden NumPy dizisi (array)

print(liste)
print(dizi)
print(type(dizi))
````

### Liste mi, dizi mi?
Aynı işlem listede ve dizide farklı sonuç verir:

````python
liste = [1, 2, 3]
dizi = np.array([1, 2, 3])

print(liste * 2)     # [1, 2, 3, 1, 2, 3] → liste tekrarlanır
print(dizi * 2)      # [2 4 6]            → her eleman 2 ile çarpılır
print(liste + liste) # [1, 2, 3, 1, 2, 3] → listeler uç uca eklenir
print(dizi + dizi)   # [2 4 6]            → eleman eleman toplanır

# Sınıfta vurgulanacak nokta: NumPy işlemi döngü yazmadan TÜM elemanlara uygular
# (vektörel işlem). Listede aynı iş için for döngüsü gerekirdi.
````

### Dizi oluşturmanın hızlı yolları ve özellikleri

````python
a = np.arange(0, 10, 2)      # range gibi: 0'dan 10'a (hariç) 2'şer
b = np.zeros(5)               # 5 tane 0
c = np.ones((2, 3))           # 2 satır, 3 sütun 1

print(a)
print(b)
print(c)

print("Boyut (shape):", c.shape)    # (satır, sütun)
print("Eksen sayısı (ndim):", c.ndim)
print("Veri tipi (dtype):", a.dtype)
````

### 2 boyutlu diziler: Bir görüntü aslında bir sayı tablosudur
Aşağıdaki 5×5'lik dizi, siyah-beyaz bir **artı (+)** işaretidir. 0 = siyah, 255 = beyaz.

````python
goruntu = np.array([
    [0,   0,   255, 0,   0],
    [0,   0,   255, 0,   0],
    [255, 255, 255, 255, 255],
    [0,   0,   255, 0,   0],
    [0,   0,   255, 0,   0],
])

print(goruntu.shape)     # (yükseklik, genişlik)
print(goruntu[2, 0])     # 3. satır, 1. sütun
print(goruntu[2])        # 3. satırın tamamı
print(goruntu[:, 2])     # 3. sütunun tamamı  (: → "hepsi")
````

### Vektörel işlemler ve toplama fonksiyonları
Bir diziye yapılan işlem **tüm elemanlara** uygulanır. `sum`, `mean`, `max`, `min` ile özet çıkarırız.
`argmax()` en büyük değerin **indeksini** (sırasını) verir; sınıflandırma modellerinde tahmin edilen sınıfı bulmak için kullanılır.

````python
skorlar = np.array([0.91, 0.45, 0.82, 0.66, 0.97])

print(skorlar * 100)          # hepsini yüzdeye çevir
print(skorlar.mean())         # ortalama
print(skorlar.max())          # en büyük değer
print(skorlar.argmax())       # en büyük değerin indeksi

# Boolean maske: koşulu sağlayanları seç
print(skorlar > 0.8)          # her eleman için True/False
print(skorlar[skorlar > 0.8]) # sadece True olanlar
print((skorlar > 0.8).sum())  # kaç tane True var?
````

### axis (eksen)
3 öğrencinin 2 sınavdan aldığı notlar bir tabloda duruyor. `axis=0` **sütun sütun** (aşağı doğru), `axis=1` **satır satır** (yana doğru) hesaplar.

````python
notlar = np.array([
    [70, 80],   # 1. öğrenci
    [90, 60],   # 2. öğrenci
    [50, 100],  # 3. öğrenci
])

print(notlar.sum())          # 450             → tüm elemanlar
print(notlar.sum(axis=0))    # [210 240]       → her SINAVIN toplamı (sütunlar)
print(notlar.mean(axis=1))   # [75. 75. 75.]   → her ÖĞRENCİNİN ortalaması (satırlar)
````

### Eksik ve tekrarlı veri
Gerçek veri setleri hiçbir zaman temiz gelmez. Geçen haftayı hatırla: **"Çöp girerse, çöp çıkar."**

| Komut | Ne yapar? |
|---|---|
| `df.isna().sum()` | Her sütunda kaç eksik değer (`NaN`) var? |
| `df.duplicated().sum()` | Kaç satır birebir tekrar ediyor? |
| `df.drop_duplicates()` | Tekrarlı satırları siler |
| `df.dropna(subset=["sutun"])` | O sütunu eksik olan satırları siler |
| `df.fillna({"sutun": deger})` | Eksikleri verilen değerle doldurur |

> Bu komutlar **yeni bir tablo döndürür**, orijinali değiştirmez. Sonucu bir değişkene atamayı unutma!

````python
print(df.isna().sum())
print()
print("Tekrarlı satır sayısı:", df.duplicated().sum())
````

### dropna'ya dikkat
`dropna()` tek başına, **herhangi bir** sütunu eksik olan tüm satırları siler. Sadece belirli bir sütuna bakmak için `subset` kullanılır:

````python
print(len(df))                             # 16
print(len(df.drop_duplicates()))           # 15  → tekrar eden 9 numaralı ilan silindi
print(len(df.dropna()))                    # 12  → eksiği olan 4 satır (7, 11, 13, 15) silindi
print(len(df.dropna(subset=["fiyat"])))    # 14  → sadece fiyatı eksik 2 satır silindi

# Sınıfta vurgulanacak nokta: dropna() tek başına çok şey silebilir.
# Bina yaşı eksik diye bir evi atmak yerine eksik değeri doldurmak daha mantıklı.
````

3. `bina_yasi` eksiklerini, `bina_yasi` sütununun **medyanıyla** doldur. *(Medyan: `df["bina_yasi"].median()`)*
