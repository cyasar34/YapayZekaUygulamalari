## 2. Metinleri Biçimlendirme (f-string)
Değişkenleri metin içine gömmek için `f"..."` kullanırız. JS'teki `` `Merhaba ${isim}` `` ile aynı mantık.

````python
isim = "Çisem"
guven = 0.87346

print(f"Merhaba {isim}!")
print(f"Güven: {guven}")              # ham hali
print(f"Güven: {guven:.2f}")          # virgülden sonra 2 basamak
print(f"Güven: %{guven * 100:.1f}")   # yüzdeye çevir, 1 basamak
print(f"Sınıf: {'kedi'.upper()}")     # .upper() metni büyük harfe çevirir
````

### for — kaç kez döneceği belliyse

````python
for sayi in range(1, 6):     # 1'den 5'e kadar (6 dahil değil)
    print(sayi)

dersler = ["Python", "Makine Öğrenmesi", "Derin Öğrenme", "Görüntü İşleme"]
for ders in dersler:
    print(f"Bu dönem işleyeceğimiz konu: {ders}")
````

### enumerate ve .items()

````python
# enumerate: elemanla birlikte sıra numarasını da verir
for sira, ders in enumerate(dersler, start=1):
    print(f"{sira}. {ders}")

# .items(): sözlükte anahtar ve değeri birlikte gezer
model_bilgisi = {"ad": "KediKopekNet", "katman": 5, "dogruluk": 0.92}
for anahtar, deger in model_bilgisi.items():
    print(f"{anahtar} → {deger}")
````

### while — koşul sağlandığı sürece

````python
epoch = 1
while epoch <= 3:
    print(f"Epoch {epoch} eğitiliyor...")
    epoch += 1          # koşulu değiştirmeyi unutma! (epoch = epoch + 1)
print("Eğitim bitti!")
````

### break ve continue

````python
skorlar = [0.9, 0.85, -1, 0.7, 0.3, 0.95]

for s in skorlar:
    if s < 0:
        continue        # geçersiz değer: bu turu atla, sonrakine geç
    if s < 0.5:
        print("Düşük skor bulundu, duruyorum.")
        break           # döngüyü tamamen bitir
    print(s)
````

### Varsayılan Parametre
Bir parametreye varsayılan değer verirsek, çağırırken o argümanı yazmak zorunlu olmaz.

````python
def tahmin_mesaji(sinif, guven, esik=0.8):
    if guven >= esik:
        return f"Bu bir {sinif}."
    return f"Emin değilim, {sinif} olabilir."

print(tahmin_mesaji("kedi", 0.93))            # esik = 0.8 (varsayılan)
print(tahmin_mesaji("kedi", 0.6))
print(tahmin_mesaji("kedi", 0.6, esik=0.5))   # esik'i değiştirdik
````

### print mi, return mü?
Garson örneğini hatırla: `print` sonucu sadece ekrana gösterir, `return` sonucu teslim eder.

````python
def topla_print(a, b):
    print(a + b)

def topla_return(a, b):
    return a + b

x = topla_print(2, 3)
y = topla_return(2, 3)
print(x, y)

# Çıktı:
# 5          ← topla_print kendi içinde yazdırdı
# None 5     ← ama bir şey teslim etmediği için x'in içi boş (None)
````
